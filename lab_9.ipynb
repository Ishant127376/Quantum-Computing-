```python
# ==============================
# LAB 9 – EXPERIMENT 1
# Quantum Key Distribution (Basic Simulation)
# ==============================

!pip install qiskit
!pip install qiskit-aer

import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import Aer

print("===== EXPERIMENT 1: QKD ====")

n = 10  # number of bits

# Alice's random bits and bases
alice_bits = np.random.randint(2, size=n)
alice_bases = np.random.randint(2, size=n)

# Bob's random bases
bob_bases = np.random.randint(2, size=n)

sim = Aer.get_backend('aer_simulator')

results = []

for i in range(n):
    qc = QuantumCircuit(1,1)

    # Alice encoding
    if alice_bits[i] == 1:
        qc.x(0)
    if alice_bases[i] == 1:
        qc.h(0)

    # Bob decoding
    if bob_bases[i] == 1:
        qc.h(0)

    qc.measure(0,0)

    result = sim.run(qc, shots=1).result()
    measured = int(list(result.get_counts().keys())[0])
    results.append(measured)

# Key generation (matching bases only)
key = []
for i in range(n):
    if alice_bases[i] == bob_bases[i]:
        key.append(results[i])

print("Alice bits: ", alice_bits)
print("Alice bases:", alice_bases)
print("Bob bases:  ", bob_bases)
print("Shared key: ", key)

print("===== DONE ====")
```

```python
# ==============================
# LAB 9 – EXPERIMENT 2
# BB84 Protocol
# ==============================

import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import Aer

print("===== EXPERIMENT 2: BB84 =====")

n = 10

alice_bits = np.random.randint(2, size=n)
alice_bases = np.random.randint(2, size=n)
bob_bases = np.random.randint(2, size=n)

sim = Aer.get_backend('aer_simulator')

bob_results = []

for i in range(n):
    qc = QuantumCircuit(1,1)

    # Alice encodes
    if alice_bits[i] == 1:
        qc.x(0)
    if alice_bases[i] == 1:
        qc.h(0)

    # Bob measures
    if bob_bases[i] == 1:
        qc.h(0)

    qc.measure(0,0)

    result = sim.run(qc, shots=1).result()
    measured = int(list(result.get_counts().keys())[0])
    bob_results.append(measured)

# Key comparison
key = []
for i in range(n):
    if alice_bases[i] == bob_bases[i]:
        key.append(bob_results[i])

print("Alice bits: ", alice_bits)
print("Alice bases:", alice_bases)
print("Bob bases:  ", bob_bases)
print("Final key:  ", key)

print("===== DONE =====")
```

```python
# ==============================
# LAB 9 – EXPERIMENT 3
# Grover's Search Algorithm
# ==============================

from qiskit import QuantumCircuit
from qiskit_aer import Aer

print("===== EXPERIMENT 3: GROVER =====")

n = 2  # number of qubits

qc = QuantumCircuit(n, n)

# Step 1: Superposition
qc.h([0,1])

# Step 2: Oracle (mark |11>)
qc.cz(0,1)

# Step 3: Diffusion operator
qc.h([0,1])
qc.x([0,1])
qc.cz(0,1)
qc.x([0,1])
qc.h([0,1])

# Measurement
qc.measure([0,1],[0,1])

print(qc.draw())

sim = Aer.get_backend('aer_simulator')
result = sim.run(qc, shots=1024).result()

print("Output:", result.get_counts())

print("===== DONE =====")
```
